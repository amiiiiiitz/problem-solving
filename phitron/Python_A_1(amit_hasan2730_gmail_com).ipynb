```python
lst = list(map(int,input().split()))
#print(type(lst[0]), type(lst))

newLst = []
for n in lst:
  if n % 4 == 0 and n % 8 != 0:
    newLst.append(n)
print(newLst)
```

Answer to the qs no 2

```python
def compute_ratio(x, y):
  try:
   return (x + y) / (x * y)
  except Exception:
    return ("Error: Division not possible")

num = list(map(int,input().split()))
returnedValue = compute_ratio(num[0], num[1])

print(returnedValue)
```

Answer to the qs no 3

```python
with open("./sample_data/data.txt","r") as file:
  content = list(map(str.split, list(map(str.strip,file.readlines()))))

#print(content)

maxLenStr = ""
for line in content:
  for word in line:
    if len(word) >= len(maxLenStr):
      maxLenStr = word

print(maxLenStr)


```

Answer to the qs no 5

```python
word = input().lower()

# print(word)

y = len(word) - 1

flag = True

for i in range(len(word) // 2):
  if word[i] != word[y]:
    flag = False
    break
  y -= 1

if flag:
  print("Palindrome")
else:
  print("Not Palindrome")
```
