### 1. Celsius to Fahrenheit Conversion

You have a list of temperatures recorded in Celsius.

Your task is to:

1. Create a NumPy array from the given list.
2. Convert all temperatures from Celsius to Fahrenheit using the formula: \\(F = C \times 1.8 + 32\\).
3. Print the new Fahrenheit array.


```python
temp = [20,30,22,34,41,18]

import numpy as np

arr = np.array(temp)
arr = arr * 1.8 + 32
print(arr)
```

## 2. Lap Times Conversion

You have a Python list containing the lap times (in seconds) for an athlete.

### Tasks

1. Convert the list into a NumPy array.
2. Print the array's shape, size, and data type.
3. The athlete wants to see their times in minutes. Create a new array by dividing all the lap times by 60.
4. Print the new array of lap times in minutes.


```python
lapTime = [200,500,100,60,400]

import numpy as np

arr = np.array(lapTime)

print(arr.shape)
print(arr.size)
print(arr.dtype)

arr2 = arr / 60
print(arr2)
```
