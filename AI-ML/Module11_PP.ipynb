1 . Create a 1D, 2D, and 3D NumPy array and print their dimensions and shapes.

```python
import numpy as np
ar1 = np.array([1,2,3,4])
print(ar1.ndim)
print(ar1.shape)

ar2= np.array([[1,2,3,4],
                [1,2,3,4]])
print(ar2.ndim)
print(ar2.shape)

ar3 = np.array([[[1,2,3],
                 [1,2,3],
                 [1,2,3]]])

print(ar3.ndim)
print(ar3.shape)
ar4 = np.array([[[1,2,3]],
                [[1,2,3]],
                [[1,2,3]]])
print(ar4.ndim)
print(ar4.shape)
```

2 . Create:


*   an array of zeros of shape (3,3)
*   an array of ones of shape (2,5)
*   an array filled with 7






```python
arrZeros = np.zeros((3,3),dtype=np.int64)
print(arrZeros)

arrOnes = np.ones((2,5),dtype=np.int64)
print(arrOnes)

arrSevens = np.full((4,4), 7)
print(arrSevens)
```
