**1. Countplot এবং Barplot এর মধ্যে পার্থক্য কী? কখন কোনটা use করবে?**

**১. Countplot:** এটি শুধুমাত্র একটি ক্যাটাগরিকাল ভ্যারিয়েবলের প্রতিটি ক্যাটাগরির ফ্রিকোয়েন্সি বা সংখ্যা (Count) দেখানোর জন্য ব্যবহার করা হয়। এখানে Y-অক্ষে সবসময় ডেটার ফ্রিকোয়েন্সি বা সংখ্যা থাকে।
**২. Barplot:** এটি একটি ক্যাটাগরিকাল ভ্যারিয়েবলের সাপেক্ষে একটি নিউমেরিক্যাল ভ্যারিয়েবলের গড় (Mean) বা অন্য কোনো এগ্রিগেট মান দেখানোর জন্য ব্যবহৃত হয়।

* **কখন ব্যবহার করবেন:**
  * যখন আপনি জানতে চান 'টাইটানিকে মোট কতজন পুরুষ এবং নারী যাত্রী ছিলেন' (শুধুমাত্র গণনা), তখন **Countplot** ব্যবহার করবেন।
  * যখন আপনি জানতে চান 'পুরুষ এবং নারী যাত্রীদের গড় টিকিট ভাড়া কত ছিল' (ক্যাটাগরি বনাম সংখ্যাগত মান), তখন **Barplot** ব্যবহার করবেন।

**2. Scatterplot দেখে কীভাবে বুঝবে দুটো variable এর মধ্যে সম্পর্ক আছে কিনা?**

Scatterplot-এ প্রতিটি ডেটা পয়েন্টের অবস্থান দেখে তাদের মধ্যে সম্পর্ক নির্ধারণ করা যায়:
১. **পজিটিভ সম্পর্ক (Positive Correlation):** যদি পয়েন্টগুলো বাম-নিচ থেকে ডান-উপরে বিন্যস্ত থাকে (অর্থাৎ একটি বাড়লে অন্যটিও বাড়ে)।
২. **নেগেটিভ সম্পর্ক (Negative Correlation):** যদি পয়েন্টগুলো বাম-উপর থেকে ডান-নিচে বিন্যস্ত থাকে (অর্থাৎ একটি বাড়লে অন্যটি কমে)।
৩. **কোনো সম্পর্ক নেই (No Correlation):** পয়েন্টগুলো যদি কোনো নির্দিষ্ট প্যাটার্ন ছাড়া এলোমেলোভাবে ছড়িয়ে ছিটিয়ে থাকে।

**3. Y Data Profiling কেন manually EDA করার চেয়ে সুবিধাজনক? কী কী information এক জায়গায় পাওয়া যায়?**

ম্যানুয়ালি প্রতিটি কলামের ডিস্ট্রিবিউশন, মিসিং ভ্যালু এবং কোরিলেশন বের করতে অনেক লাইনের কোড লিখতে হয় এবং অনেক সময় লাগে। **YData Profiling** মাত্র এক লাইনের কোড দিয়ে পুরো ডেটাসেটের একটি পূর্ণাঙ্গ ইন্টারঅ্যাক্টিভ HTML রিপোর্ট তৈরি করে দেয়।

**একই জায়গায় যেসব তথ্য পাওয়া যায়:**
* **Dataset Overview:** ডেটাসেটের মোট রো, কলাম, মেমোরি সাইজ এবং মিসিং ভ্যালুর শতকরা হার।
* **Variables Detail:** প্রতিটি কলামের টাইপ, ইউনিক ভ্যালু, ডিস্ট্রিবিউশন (Histogram) এবং স্ট্যাটিস্টিকস (Mean, Median, Min, Max)।
* **Correlations:** ভ্যারিয়েবলগুলোর নিজেদের মধ্যকার সম্পর্কের হিটম্যাপ (Heatmap)।
* **Missing Values:** কোন কোন কলামে কতগুলো মিসিং ডেটা আছে তার ভিজ্যুয়াল চার্ট।

**4. নিচের code টি একটি Countplot তৈরি করবে যেখানে passenger class অনুযায়ী count দেখাবে এবং Gender দিয়ে আলাদা করবে।**

*Hint: x = ticket class column, hue = gender column*

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("titanic")

sns.countplot(data=df, x="class", hue="sex")

plt.title("Passenger Class by Gender")
plt.show()
```

**5. নিচের code টি City অনুযায়ী average Salary দেখাবে।**

*Hint: Cat-Num relationship দেখানোর plot, x = category, y = number*

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

df = pd.DataFrame({
    "City": ["Dhaka", "Dhaka", "CTG", "CTG", "Sylhet"],
    "Salary": [50000, 60000, 45000, 55000, 40000]
})

sns.barplot(data=df, x="City", y="Salary")

plt.title("Average Salary by City")
plt.show()
```

**6. নিচের code টি Age column এর smooth distribution curve দেখাবে।**

*Hint: Histogram এর smooth version যে plot দেখায়*

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("titanic")

sns.histplot(data=df, x="age", kde=True)

plt.title("Age Distribution")
plt.show()
```

**7. নিচের code টি Height এবং Weight এর মধ্যে সম্পর্ক দেখাবে।**

*Hint: দুটো numerical variable এর মধ্যে relationship দেখানোর plot*

**Bonus 1: নিচের code টি দিয়ে Titanic dataset এর Class এবং Fare এর মধ্যে সম্পর্ক দেখাতে একটি Boxplot তৈরি করো।**

*Hint: x = categorical column, y = numerical column*

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("titanic")

# Boxplot তৈরির জন্য সঠিক ফাংশন এবং কলাম বসাও
sns.boxplot(data=df, x="class", y="fare")

plt.title("Fare Distribution by Class")
plt.show()
```


**Bonus 2: নিচের code টি দিয়ে Titanic dataset-এ Gender (Sex) অনুযায়ী survival rate (বা বেঁচে থাকার পারসেন্টেজ) বের করো।**




```python
import seaborn as sns
import pandas as pd

df = sns.load_dataset("titanic")

# gender অনুযায়ী survival rate বের করো
gender_split = df.groupby('sex')['survived'].mean()
# gender_split = df.groupby('sex')['survived'].value_counts(normalize=True)

print(gender_split)
```


**Bonus 3:নিচের code টি দিয়ে Titanic dataset-এ Passenger Class ('class') অনুযায়ী টিকেটের গড় মূল্য (average fare) বের করো।**

*Hint: `groupby` ব্যবহার করো 'class' কলামের ওপর এবং 'fare' কলামের `mean()` বের করো।*


```python
import seaborn as sns
import pandas as pd

df = sns.load_dataset("titanic")

# class অনুযায়ী average fare বের করার জন্য সঠিক ফাংশন ও কলাম বসাও
class_fare = df.groupby('class')['fare'].mean()

print(class_fare)
```
