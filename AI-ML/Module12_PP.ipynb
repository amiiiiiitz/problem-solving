1. Create a Pandas Series from the list [10, 20, 30, 40, 50]. Print the Series, its index, and data type. Then create a DataFrame with columns "Name", "Age", and "City" containing at least 4 rows of data.


```python
import pandas as pd
import numpy as np

# Part 1: Create a Pandas Series
lst = [10, 20, 30, 40, 50]

series = pd.Series(lst)

print("Series:")
print(series)

print("\nIndex:")
# print(series.index)
index = list(series.index)
print(index)

print("\nData Type:")
print(series.dtype)


# Part 2: Create a Pandas DataFrame
newDF = pd.DataFrame(
    [
        ["Amit Hasan", 23, "Dhaka"],
        ["Shaikot", 30, "Noakhali"],
        ["Rahim", 22, "Chittagong"],
        ["Karim", 25, "Sylhet"]
    ],
    columns=["Name", "Age", "City"]
)

print("\nDataFrame:")
print(newDF)
```

4.. Using .loc[] and .iloc[], show the first row, the second column, and a 2×2 block    from the middle of your DataFrame.

```python
# 1. Access the first row using .loc[]
# .loc[] uses index labels; the ending label is included.
print("First row using loc:")
print(df.loc[0:0])


# 2. Access the first row using .iloc[]
# .iloc[] uses integer positions; the ending position is excluded.
print("\nFirst row using iloc:")
print(df.iloc[0:1])


# 3. Access the second column using .loc[]
# Select all rows (:) and the column named "Price".
print("\nSecond column using loc:")
print(df.loc[:, "Price"])


# 4. Access the second column using .iloc[]
# : means all rows; 1:2 selects the column at position 1.
print("\nSecond column using iloc:")
print(df.iloc[:, 1:2])


# 5. Access a 2×2 block using .loc[]
# Select rows with labels 1 and 2,
# and columns from "Product" through "Price", inclusive.
print("\n2×2 block using loc:")
print(df.loc[1:2, "Product":"Price"])


# 6. Access a 2×2 block using .iloc[]
# Select rows at positions 1 and 2,
# and columns at positions 0 and 1.
print("\n2×2 block using iloc:")
print(df.iloc[1:3, 0:2])
```


6.Create a DataFrame of employees with columns "Name", "Department", and "Salary". Sort it by "Salary" in ascending order and by "Name" in descending order.


```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Amit", "Rahim", "Karim", "Nadia"],
    "Department": ["IT", "HR", "Finance", "IT"],
    "Salary": [50000, 40000, 45000, 55000]
})

print(df)
df.sort_values(['Salary','Name'], ascending=[1,0])
```

7.Create a DataFrame of 8 students with columns "Name" and "Score". Filter and print students with Score > 60 and those with Score < 40.
