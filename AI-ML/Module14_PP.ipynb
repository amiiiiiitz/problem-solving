```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'A': [1, 2, np.nan, 4, 5],
    'B': [5, np.nan, np.nan, None, 10],
    'C': ['a', 'b', 'c', None, 'e']
})

print(df)

# 1. Replace null values based on conditions
df1 = df.copy()
df1.fillna({'A': 0, 'B': 99, 'C': 'Unknown'}, inplace=True)
print(df1)

# 2. Forward fill only column B
df2 = df.copy()
df2['B'] = df2['B'].ffill()
print(df2)

# 3. Fill numeric columns with mean and string columns with Unknown
df3 = df.copy()

df3['A'] = df3['A'].fillna(df3['A'].mean())
df3['B'] = df3['B'].fillna(df3['B'].mean())
df3['C'] = df3['C'].fillna('Unknown')

print(df3)
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'A': [1, 2, np.nan, 4, 5],
    'B': [5, np.nan, np.nan, None, 10],
    'C': ['a', 'b', 'c', None, 'e']
})

print(df)

# 1. Drop rows where column A has missing values
df1 = df.copy()
df1 = df1.dropna(subset=['A'])
print(df1)

# 2. Drop rows with less than 2 non-null values
df2 = df.copy()
df2 = df2.dropna(thresh=2)
print(df2)

# 3. Drop rows where both A and B have missing values
df3 = df.copy()
df3 = df3.dropna(subset=['A', 'B'], how='all')
print(df3)
```

```python
import pandas as pd
import numpy as np

# Create a sample DataFrame
df = pd.DataFrame({
    'Category': ['A', 'B', 'A', 'B', 'A', 'B'],
    'Value1': [10, 20, 30, 40, 50, 60],
    'Value2': [100, 200, 300, 400, 500, 600]
})

# Basic groupby and aggregation
print("Mean of Value1 for each Category:")
print(df.groupby('Category')['Value1'].mean())

print("Max of Value2 for each Category:")
print(df.groupby('Category')['Value2'].max())

print("Sum of Value1 for each Category:")
print(df.groupby('Category')['Value1'].sum())

```
