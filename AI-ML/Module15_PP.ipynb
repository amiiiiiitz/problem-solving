1 Add New Column and Save File.

You have a DataFrame with column Price.
* Create a new column Discounted_Price with (10% discounted price).
* Save the modified DataFrame as discounted_products.csv.


```python
import pandas as pd
df = pd.read_csv('Practice.csv')

# display(df)

df['Discounted_Price'] = df['Price'] * .90

df.to_csv('discounted_products.csv',  index = False)

display(pd.read_csv('discounted_products.csv'))
```

2 Check Duplicates and Remove Them

Column: EmployeeID

Tasks:
* Count duplicate entries


* Remove duplicates


* Display cleaned DataFrame


```python
display(df)

print(df.duplicated(subset = ['EmployeeID']).sum())
print(len(df['EmployeeID']))

df.drop_duplicates(subset = ['EmployeeID'], inplace = True)

print(len(df['EmployeeID']))

display(df)
```

3 Handle Missing Values

Columns: Age, Salary

Tasks:
* Count missing values


* Fill missing Age with mean age


* Drop rows where Salary is missing


```python
df3 = pd.read_csv('Practice.csv')
display(df3)
print('\n')

missing_count_age = df3['Age'].isnull().sum()
print(missing_count_age)
missing_count_salary = df3['Salary'].isnull().sum()
print(missing_count_salary)

df3['Age'] = df3['Age'].fillna(df3['Age'].mean())

display(df3)
print('\n')

df3.dropna(subset = ['Salary'], inplace = True)

display(df3)
```

4 Apply Statistical Functions

Columns: Marks1, Marks2, Marks3

 Tasks:

* Calculate mean, max for each subject
* Compute total marks per student


* Find standard deviation of total marks


```python
print(df3['Marks1'].mean())
print(df3['Marks2'].mean())
print(df3['Marks3'].mean())

print('\n')

print(df3['Marks1'].max())
print(df3['Marks2'].max())
print(df3['Marks3'].max())

total_marks = df3['Marks1'] + df3['Marks2'] + df3['Marks2']

display(total_marks)

display(total_marks.std())
```
