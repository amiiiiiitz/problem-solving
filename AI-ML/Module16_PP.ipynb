```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

1 Using sales, draw a lineplot of date vs sales.

```python
sales = pd.read_csv('sales.csv')
sns.lineplot(data= sales , x = 'date' , y = 'sales',errorbar=None)
```

2 From students, plot hours_studied vs marks and color by gender.

```python
students = pd.read_csv("students.csv")
# group_by_gender = students.groupby('gender').size()
# display(group_by_gender)

sns.lineplot(data= students , x = 'hours_studied' , y = 'marks',errorbar=None,hue='gender')
```

3 Create a grid of scatterplots (hours_studied vs marks) faceted by subject.

```python
sns.relplot(
    data=students,
    x='hours_studied',
    y='marks',
    col='subject',
    kind='scatter',
)
```

4 Plot a histogram of marks with a KDE overlay.

```python
sns.histplot(data = students , x ='marks',hue='gender', kde = True, element='step')
```

5 Show the KDE curve for hours_studied.

```python
sns.kdeplot(data = students, x ='hours_studied',fill=True, hue="gender")
plt.show()
```

6 Display the count of records per subject

```python
sns.countplot(data = students , x='subject',hue='gender')
```

7 Draw a regression line for hours_studied and marks.

```python
sns.lmplot(data = students , x ='hours_studied', y = 'marks', hue='gender')
```

8 Create a pairplot for students' dataset.

```python
students_pair = students[['age','subject', 'previous_gpa', 'attendance_percentage']]
# display(students_pair)

# Pairplot only plots numerical columns.
# Here, all columns except 'subject' are numerical, so they are included in the plots.
# Since 'subject' is categorical, we use it as 'hue' to distinguish different subjects using colors.
sns.pairplot(data=students_pair,hue='subject')

plt.show()
```
