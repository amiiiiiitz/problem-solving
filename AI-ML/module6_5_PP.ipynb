```python
def introduce(name, age):
  print(f"My name is {name} and I'm {age} years old")

name = input("enter your name: ")
age = int(input("enter your age: "))
introduce(name, age)
```

```python
def batch_generator(lst, batchSize):
  for i in range(0,len(lst), batchSize):
    yield lst[i:i+batchSize]

lst = input("enter your list: ").split()
print(lst)
print(type(lst[0]))

g = batch_generator(lst, 3)

print(next(g))
print(next(g))
print(next(g))
```

```python
def batch_generator(lst, batchSize):
  for i in range(0,len(lst), batchSize):
    yield lst[i:i+batchSize]

lst = input("enter your list: ").split()
print(lst)
print(type(lst[0]))
batchSize = int(input())

g = batch_generator(lst, batchSize)

for batch in g:
  print(batch)
```

```python
def power(base, exponent = 2):
  return base ** exponent

lst = list(map(int,input().split()))
print(power(lst[0], lst[1])) if len(lst) > 1 else print(power(lst[0]))
```

```python
lst = [1,2,3,4,5,6]

sq_ev = list(map(lambda x: x**2, list(filter(lambda x: x % 2 == 0, lst))))
print(sq_ev)
```
