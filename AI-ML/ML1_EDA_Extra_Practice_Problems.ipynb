```python
# Import the libraries and load the dataset
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('titanic_data_updated (1).csv')
```

### Problem 1: Explore the Dataset

1. Display the first 10 rows of the dataset.
2. Display 10 random rows.
3. Find the total number of rows and columns.
4. Print the number of rows and the number of columns separately.

```python
# Problem 1: Explore the Dataset

# 1. Display the first 10 rows of the dataset.
print("--- First 10 Rows ---")
display(df.head(10))

# 2. Display 10 random rows.
print("\n--- 10 Random Rows ---")
display(df.sample(10))

# 3. Find the total number of rows and columns.
shape = df.shape
print(f"\nTotal Shape: {shape}")

# 4. Print the number of rows and the number of columns separately.
print(f"Number of Rows: {shape[0]}")
print(f"Number of Columns: {shape[1]}")
```

### Problem 2: Check the Dataset Structure

1. Display the dataset information, including column names, data types, and non-null values.
2. Generate a statistical summary of the numerical columns.
3. Identify which columns contain numerical data and which contain non-numerical data.

```python
# Problem 2: Check the Dataset Structure

# 1. Display the dataset information
print("--- Dataset Information ---")
df.info()

# 2. Generate a statistical summary of the numerical columns
print("\n--- Statistical Summary ---")
display(df.describe())

# 3. Identify which columns contain numerical and non-numerical data
numerical_cols = df.select_dtypes(include=[np.number]).columns.tolist()
non_numerical_cols = df.select_dtypes(exclude=[np.number]).columns.tolist()
print(f"\nNumerical columns: {numerical_cols}")
print(f"Non-numerical columns: {non_numerical_cols}")
```

### Problem 3: Find Missing Values

1. Count the missing values in each column.
2. Display only the columns that contain missing values.
3. Identify the column with the highest number of missing values.

```python
# Problem 3: Find Missing Values

# 1. Count the missing values in each column
missing_values = df.isnull().sum()
print("--- Missing Values Count ---")
print(missing_values)

# 2. Display only the columns that contain missing values
print("\n--- Columns with Missing Values ---")
print(missing_values[missing_values > 0])

# 3. Identify the column with the highest number of missing values
highest_missing_col = missing_values.idxmax()
highest_missing_val = missing_values.max()
print(f"\nColumn with the highest number of missing values: {highest_missing_col} ({highest_missing_val} missing values)")
```

### Problem 4: Check Duplicate Rows

1. Count the total number of duplicate rows.
2. Create a separate DataFrame without duplicate rows.
3. Display the number of rows before and after removing duplicates.

```python
# Problem 4: Check Duplicate Rows

# 1. Count the total number of duplicate rows
duplicate_count = df.duplicated().sum()
print(f"Total duplicate rows: {duplicate_count}")

# 2. Create a separate DataFrame without duplicate rows
df_cleaned = df.drop_duplicates()

# 3. Display the number of rows before and after removing duplicates
print(f"Rows before: {df.shape[0]}")
print(f"Rows after: {df_cleaned.shape[0]}")
```

### Problem 5: Analyze Passenger Survival

1. Count how many passengers survived and how many did not survive.
2. Calculate the percentage for each group.
3. Create a Seaborn count plot for the `Survived` column.

```python
# Problem 5: Analyze Passenger Survival

# 1. Count how many passengers survived and how many did not
survival_counts = df['Survived'].value_counts()
print("--- Survival Counts ---")
print(survival_counts)

# 2. Calculate the percentage for each group
survival_percentages = df['Survived'].value_counts(normalize=True) * 100
print("\n--- Survival Percentages ---")
print(survival_percentages)

# 3. Create a Seaborn count plot for the Survived column
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='Survived')
plt.title('Passenger Survival Counts')
plt.xlabel('Survived')
plt.ylabel('Count')
plt.show()
```

### Problem 6: Analyze Categorical Features

For each of these columns — `Pclass`, `Sex`, and `Embarked`:

1. Count the number of records in each category.
2. Calculate the percentage of each category.
3. Create a separate Seaborn count plot.


```python
# Problem 6: Analyze Categorical Features

columns_to_analyze = ['Pclass', 'Sex', 'Embarked']

for col in columns_to_analyze:
    print(f"\n=== Analysis for {col} ===")
    # 1. Count the number of records
    counts = df[col].value_counts(dropna=False)
    print("Counts:")
    print(counts)

    # 2. Calculate the percentage
    percentages = df[col].value_counts(normalize=True, dropna=False) * 100
    print("Percentages:")
    print(percentages)

    # 3. Create count plot
    plt.figure(figsize=(6, 4))
    sns.countplot(data=df, x=col)
    plt.title(f'Count Plot of {col}')
    plt.xlabel(col)
    plt.ylabel('Count')
    plt.show()
```

### Problem 7: Explore the Fare Distribution

1. Create a histogram of the `Fare` column using 50 bins.
2. Create a KDE plot for the `Fare` column.
3. Create a box plot for the `Fare` column.


```python
# Problem 7: Explore the Fare Distribution

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# 1. Create a histogram of the Fare column using 50 bins
sns.histplot(data=df, x='Fare', bins=50, ax=axes[0], kde=False)
axes[0].set_title('Fare Histogram (50 Bins)')

# 2. Create a KDE plot for the Fare column
sns.kdeplot(data=df, x='Fare', ax=axes[1], fill=True)
axes[1].set_title('Fare KDE Plot')

# 3. Create a box plot for the Fare column
sns.boxplot(data=df, x='Fare', ax=axes[2])
axes[2].set_title('Fare Box Plot')

plt.tight_layout()
plt.show()
```
